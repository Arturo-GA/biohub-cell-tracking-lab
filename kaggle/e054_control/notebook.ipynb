# Pure public tight55 control
Recompute all test images and export the pure control as submission.csv. No test prediction cache, no training annotations, no complementary bridge stage.

```python
import base64,hashlib,io,os,subprocess,sys,zipfile
from pathlib import Path
payload=base64.b64decode('')
assert hashlib.sha256(payload).hexdigest()=='4ed765ccee8d9ff2b00339918baecd54c0c2aa3c3499d40edf5436f2b215c7a0'
package=Path('/kaggle/working/e054_control_package');package.mkdir(exist_ok=True)
with zipfile.ZipFile(io.BytesIO(payload)) as archive:
    for name in archive.namelist():
        if not (package/name).resolve().is_relative_to(package.resolve()):raise ValueError(name)
    archive.extractall(package)
env=dict(os.environ,PYTHONPATH=os.pathsep.join([str(package/'src'),str(package/'scripts')]),
         OPENBLAS_NUM_THREADS='1',MKL_NUM_THREADS='1',OMP_NUM_THREADS='2')
subprocess.run([sys.executable,'-u',str(package/'scripts/public_control_submission_runner.py'),str(package)],env=env,check=True)
```
