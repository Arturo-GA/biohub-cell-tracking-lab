```python
import base64,hashlib,io,os,subprocess,sys,zipfile
from pathlib import Path
payload=base64.b64decode('')
assert hashlib.sha256(payload).hexdigest()=='0f287fa08b45b1c5dff92f44faa7e243b04465dcf92ecf7a7d63e6bb26eb0355'
package=Path('/kaggle/working/e064_submission_package');package.mkdir(exist_ok=True)
with zipfile.ZipFile(io.BytesIO(payload)) as archive:
    for name in archive.namelist():
        if not (package/name).resolve().is_relative_to(package.resolve()):raise ValueError(name)
    archive.extractall(package)
env=dict(os.environ,PYTHONPATH=os.pathsep.join([str(package/'src'),str(package/'scripts')]),
         OPENBLAS_NUM_THREADS='1',MKL_NUM_THREADS='1',OMP_NUM_THREADS='2')
subprocess.run([sys.executable,'-u',str(package/'scripts/native_centers_submission_runner.py'),str(package)],env=env,check=True)
```
